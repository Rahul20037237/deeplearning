```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.feature_extraction.text import  CountVectorizer
from sklearn.naive_bayes import GaussianNB
import pandas as pd
import re
from sklearn.linear_model import  LogisticRegression
from sklearn.metrics import  f1_score
from sklearn.model_selection import train_test_split
from sklearn.pipeline import  Pipeline
```

```python
corpus = pd.read_json(r'/root/.cache/kagglehub/datasets/rmisra/news-category-dataset/versions/3/News_Category_Dataset_v3.json', lines=True)
corpus1=corpus[['headline','category']][:1000]
# print(corpus1.head()
print(corpus['category'].value_counts())
```
